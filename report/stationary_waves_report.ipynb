# Stationary Waves

## Exact Solution

$ u(t,x,y) = e^{i(k_x x + k_y y - \omega t)} \qquad \frac{\partial^2 u}{\partial t^2} = c^2 \nabla^2 u$

LHS: $\frac{\partial^2}{\partial t^2}(e^{i(k_x x + k_y y - \omega t)}) = -\omega^2 e^{i(k_x x + k_y y - \omega t)} \qquad$

RHS: $c^2 \Big[\frac{\partial^2}{\partial x^2}(e^{i(k_x x + k_y y - \omega t)}) + \frac{\partial^2}{\partial y^2}(e^{i(k_x x + k_y y - \omega t)})\Big]$

$= -c^2 \Big[k_x^2 (e^{i(k_x x + k_y y - \omega t)}) + k_y^2 (e^{i(k_x x + k_y y - \omega t)})\Big] = -e^{i(k_x x + k_y y - \omega t)} \Big[c^2(k_x^2 + k_y^2)\Big]$

$\implies \text{setting} \;\;\; \omega^2 = c^2 (k_x^2 + k_y^2)\;\;\;$ makes the LHS and RHS identical

## Dispersion coefficient

Inserting:

$$
u_{ij}^n = e^{i(kh(i+j)-\tilde{\omega}n \Delta t)}
$$

into

$$
\frac{u^{n+1}_{i,j} - 2u^n_{i,j} + u^{n-1}_{i, j}}{\Delta t^2} =
c^2 \left(\frac{u^n_{i+1,j} - 2u^n_{i,j} + u^n_{i-1, j}}{h^2} + \frac{u^n_{i,j+1} - 2u^n_{i,j} + u^n_{i, j-1}}{h^2}\right)
$$

Central differencing time shifts gives:

$$
u_{ij}^{n+1} = u_{ij}^n e^{-i \tilde{\omega} \Delta t}
, \qquad 
u_{ij}^{n-1} = u_{ij}^n e^{i \tilde{\omega} \Delta t}
$$

Spatial shifts gives:

$$
u_{i+1,j}^{n} = u_{i,j+1}^n = u_{ij}^n e^{i k h}
, \qquad
u_{i-1,j}^{n} = u_{i,j-1}^n = u_{ij}^n e^{-i k h}
$$

Substitute and cancel:

$$
\frac{e^{-i \tilde{\omega} \Delta t} - 2 + e^{i \tilde{\omega} \Delta t}}{\Delta t^2} = \frac{2c^2}{h^2} (e^{i k h} - 2 + e^{-i k h})
$$


Use:

$$
e^{i \theta} + e^{-i \theta} - 2 = 2\cos(\theta) - 2 = -4 \sin^2(\frac{\theta}{2})
$$

Then:

$$
-\frac{4}{\Delta t^2} \sin^2 (\frac{\tilde{\omega} \Delta t}{2}) = - \frac{8c^2}{h^2} \sin^2(\frac{kh}{2})
$$




Since $ \ C = \frac{c \Delta t}{h}$

$$
\sin^2 \Big(\frac{\tilde{\omega} \Delta t}{2} \Big) = 2 C^2 \sin^2 \Big(\frac{kh}{2}\Big)
$$

If $ \ C = \frac{1}{\sqrt{2}} \ \implies \tilde{\omega} \Delta t = k h$

$$
\Delta t = \frac{h}{\sqrt{2}c} \implies \tilde{\omega} = \frac{k h \sqrt{2} c}{h} = \sqrt{2}ck
$$

From exact solution $ \omega^2 = c^2 (2 k^2) \implies \omega = \sqrt{2}ck$ , therefore:

$$
\tilde{\omega} = \omega
$$


## Test for exact solution

The scheme uses two previous time steps, therefore both $U^0$ and $U^1$ are needed. Let the fixed spatial shape for the Dirichlet and Neumann condition respectively be:

$$
\Phi_{ij} = \sin({k x_i}) \sin({k y_j})
\qquad
\Phi_{ij} = \cos({k x_i}) \cos({k y_j})
$$

The exact solution on the grid is:

$$
U_{e}^{n} = \Phi \cos(\omega n \Delta t)
$$

At the chosen CFL=$\frac{1}{\sqrt{2}},$ $\;\;\;\omega \Delta t = kh,\;\;\;$ so the above becomes:

$$
U_{e}^{n} = \Phi \cos(nkh)
$$

So the starting values are:

$$
U_e^0 = \Phi 
\qquad
U_e^1 = \Phi \cos(kh)
$$

Check that the first numerical step produces $U^1 = \Phi \cos(kh)$. Here $\nabla_h^2$ denotes the discrete (central difference) Laplacian:

$$
U^1 = U^0 + \frac{c^2 \Delta t^2}{2} \nabla_h^2 U^0
$$

Trigonometry sum of angles identities, here $s$ can be either $\sin$ or $\cos$:

$$
s(k(x+h)) + s(k(x-h)) = 2\cos(kh)s(kx)
$$

Therefore, the second difference in one direction multiplies the sampled wave by the following:

$$
\frac{2 \cos(kh) - 2}{h^2} = -\frac{4}{h^2} \sin^2\Big(\frac{kh}{2}\Big)
$$

Because the two wave numbers are equal, the two directions add, so for the discrete Laplacian:

$$
\nabla_h^2 \Phi = - \frac{8}{h^2}\sin^2 \Big(\frac{kh}{2}\Big) \Phi
$$


Dirichlet: The relation is used at the interior nodes, the boundary values are held at zero.

Neumann: The reflected boundary stencils preserve the same relation at the boundary

Substitute into the first step formula:

$$
U^1 = \Phi + \frac{c^2 \Delta t^2}{2} \Big[-\frac{8}{h^2} \sin^2 \Big(\frac{kh}{2} \Big) \Phi \Big] = \Big[1 - 4C^2 \sin^2 \Big(\frac{kh}{2} \Big) \Big] \Phi
$$

Use $C$ and trigonometric identity to get:

$$
U^1 = \Phi \cos(kh) = U_e^1
$$

Check the steps going forward:

$$
U^n = a_n \Phi
$$

Here $a_n$ is the amplitude at step $n$. Substitute into the time-stepping formula:

$$
a_{n+1} = \Big[2 - 8 C^2 \sin^2 \Big(\frac{kh}{2} \Big) \Big]a_n - a_{n-1}
$$

Use $C$ and trigonometric identities

$$
a_n = \cos(nkh)
$$

Therefore every step is exact. This choice makes the numerical solution identical to analytical.
